```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split
```

```python
#Create a dataset:100 student record
np.random.seed(42)
```

```python
df=pd.DataFrame({
    'Student_Id':range(1,101),
    'Name': [f'Student_{i}' for i in range(1,101)],
    'Gender':np.random.choice(['Male','Female'],100),
    'Department':np.random.choice(['CSE','ECE','ME','CE'],100),
    'Marks': np.random.randint(40,101,100)
})
```

```python
df
```

```python
#To Find the length of dataset
print("Population Size",len(df))
```

```python
#About Data/Population Ispection
print("Department Distribution",df['Department'].value_counts())

#Gender Distribution
print("Gender Distribution",df['Gender'].value_counts())

#Marks Distribution
print('\n Basic Distribution of Marks',df['Marks'].describe())
```

```python
#Simple Random Sampling:Equal chance for selection for every population
sample_size= 13
simple_random_sample=df.sample(
    n=sample_size,
    random_state=42
)
```

```python
simple_random_sample
```

```python
#Visualization for simple Random Selection
plt.figure(figsize=(12,6))

plt.scatter(df['Student_Id'],df['Marks'],alpha=0.35,label='Population')

plt.scatter(simple_random_sample['Student_Id'],
            simple_random_sample['Marks'],
            s=120,label='Selected Sample'
            )
plt.xlabel("STUDENT ID")
plt.ylabel("MARKS")
plt.title("SIMPLE RANDOM SAMPLE")
plt.legend()
plt.show()
```

# **08-09-2026**

```python
#Systematic Sampling
N=len(df)
n=10
k=N//n
start=np.random.randint(0,k)
systematic_sampling = df.iloc[start : : k].head(n)
print("Sampling Interval: K=",k)
print("Starting Position: ", start)
systematic_sampling
```

```python
#Visualization of Systematic Sampling
plt.figure(figsize=(12,5))
plt.scatter(df["Student_Id"],df['Marks'],alpha=0.35,label='population')
plt.scatter(systematic_sampling["Student_Id"],systematic_sampling['Marks'],s=120,label='Selected Sample')
plt.xlabel("Student_ID")
plt.ylabel("Marks")
plt.title("Systematic Sampling")
plt.legend()
plt.show
```

```python
# Stratified sampling : strategy
sample_size=20
stratified_sample , remaining= train_test_split(df,
                                                   test_size=0.80,
                                                   stratify=df['Department'],
                                                random_state=42)
stratified_sample
```

```python
# Cluster Sampling
cluster = df['Department'].unique()
selected_clusters=np.random.choice(cluster,size=2,replace=False)
cluster_sample = df[df['Department'].isin(selected_clusters)]
print("Selected Cluster: ", selected_clusters)
print("\n Sample_Size:", len(cluster_sample))
print(cluster_sample)
```

```python
# Quota Sampling
male_sample = df[df['Gender']=='Male'].sample (n=5, random_state=42)
female_sample = df[df['Gender']=='Female'].sample (n=5, random_state=42)
quota_sample = pd.concat([male_sample , female_sample])
print(quota_sample['Gender'])
```
