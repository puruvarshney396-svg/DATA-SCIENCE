# **Using Group By Function**

```python
import pandas as pd
```

```python
df=pd.read_csv("/content/weather_by_cities.csv")
```

```python
df
```

```python
g=df.groupby("city")
```

```python
g
```

```python
for city , data in g:
  print("City",city)
  print("\n")
  print("data",data)
```

```python
g.max()
```

```python
g.min()
```

```python
g.mean(numeric_only=True)
```

```python
g.describe()
```

```python
g.size()
```

```python
import matplotlib.pyplot as plt
```

```python
g=df.groupby("city")["temperature"].mean()
plt.bar(g.index,g.values)
plt.title("Mean of Temperature")
plt.xlabel("City")
plt.ylabel("Temperature")
plt.grid(True,linestyle=":",alpha=0.5)
```

```python
df1=df.set_index(["city","day"])
g1=(df1.groupby(level=0)["temperature"].mean())
```

```python
g1.head(10)
```

```python
print(df.groupby("city",as_index=True)["temperature"].mean())
```

```python
print(df.groupby("city",sort=True)["temperature"].mean())
```

```python
print(df.groupby("city",observed=True)["temperature"].mean())
```

```python
print(df.groupby("city",dropna=True)["temperature"].mean())
```

# **Aggregate Function**

```python
df
```

```python
df.aggregate("mean",numeric_only=True)
```

```python
df.aggregate({"temperature":["min","max","mean"],
              "windspeed":["min","max"]})
```

```python
df["temperature"].aggregate("mean")
```

```python
df
df[["temperature","windspeed"]].aggregate("sum",axis=1)
```

```python
df[["temperature","windspeed"]].aggregate("sum")
```

```python
table=df.pivot_table(
    values=["temperature","windspeed"],
    index="city",
    columns="day"
)
print(table)
```

```python
table=df.pivot_table(
    values=["temperature","windspeed"],
    index="city",
    aggfunc="sum"
)
print(table)
```

```python
table=df.pivot_table(
    values=["temperature","windspeed"],
    index="city",
    aggfunc=["sum","max","min"]
)
print(table)
```
