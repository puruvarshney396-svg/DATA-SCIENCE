#  **JSON File**

# **27-10-26**

```python
# create a json file
import json
import pandas as pd
```

```python
stu = [{"Name": "Puru","Department" : "CSE","Marks":90},
       {"Name": "Tara","Department" : "IT","Marks":80},
       {"Name": "Ayush","Department" : "Robotics","Marks":89}
       ]
```

```python
print(stu)
```

```python
#Create a Json File
with open("Student.json","w",encoding = "utf-8")as file:
  json.dump(stu,file,indent=3 )
```

```python
#Open the file and read their content
with open("Student.json","r",encoding = "utf-8")as file:
  print(file.read())
```

```python
#Read Json Data
newstudent = pd.read_json("/content/Student.json")
print(newstudent)
```

```python
## Find students whose marks are greater than 80
newstudent[newstudent["Marks"] > 80]
```

```python
#Conversion of Data Frame
df= pd.DataFrame(newstudent)
print(df)
```

```python
# Convert into csv
df.to_csv("newstudent.csv",index=False)
```

```python
df=pd.read_csv("/content/newstudent.csv")
print(df)
```

# **SQL LITE**

```python
import sqlite3
conn=sqlite3.connect("student.db")
cursor=conn.cursor()
print("Database created Successfully")
```

```python
#Create a table in sqlite | Students.db
cursor.execute("""
CREATE TABLE IF NOT EXISTS student_table(
RollNo INTEGER,
NAME TEXT,
Marks INTEGER)
""")
print("Successfully Created the Table")
```

```python
#Insert the record
cursor.execute("INSERT INTO student_table VALUES(?,?,?) ",
               (101,"AMIT",80))
cursor.execute("INSERT INTO student_table VALUES(?,?,?) ",
               (102,"AAYUSH",90))
cursor.execute("INSERT INTO student_table VALUES(?,?,?) ",
               (102,"MITU",60))
conn.commit()
print("Successfully INSERTED THE RECORDS")
conn.close()
```

```python
conn=sqlite3.connect("student.db")
cursor=conn.cursor()
cursor.execute("SELECT * FROM student_table")
rows=cursor.fetchall()
for row in rows:
  print(row)

conn.close()
```
