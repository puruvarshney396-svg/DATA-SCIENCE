# **Data Wrangling (03-11-26)**

```python
import pandas as pd
```

```python
data = pd.Series(
    [85,90,56,30,40],
    index=[
        ["CSE","IT","ECE","CSE","BSC"],
        ["Amit","Aayush","Montu","Akshara","Princii"]
    ]
)
print(data)
```

```python
#First Indexing
print(data["CSE"])
```

```python
#Hierarchical Indexing
print(data["CSE"]["Akshara"])
```

```python
#Set Index
df=pd.DataFrame({
    "Student":["Puru","Puru","Ashi","Ashi","Akshara","Akshara"],
    "Subject":["Java","Python","Python","Java","Python","C++"],
    "Marks":[90,87,84,88,88,87]
})
print(df)
```

```python
df1=df.set_index(["Student","Subject"])
```

```python
print(df1)
```

```python
print(df1.reset_index())
```

```python
# Concatination Method (Row Wise)
A=pd.DataFrame({"Name":["Amit","Priya"],
               "Marks":[88,90]})
B=pd.DataFrame({"Name":["Puru","Priyanshi"],
               "Marks":[98,92]})
C=pd.concat([A,B],ignore_index=True)
print(C)
```

```python
# Concatination Method (Column wise)
A1=pd.DataFrame({"Name":["Amit","Priya"],
               "Marks":[88,90]})
B1=pd.DataFrame({"Attendance":[80,90]})
C1=pd.concat([A1,B1],axis=1)
print(C1)
```

```python
# Merge Operation
Student=pd.DataFrame({"Student_id":[101,102,103],
               "Name":["Amit","Priya","Purav"]})
Marks=pd.DataFrame({"Student_id":[101,102,104],
               "Marks":[80,90,88]})
result=pd.merge(Student,Marks,on="Student_id")
print(result)
```

```python
#inner join
pd.merge(Student,Marks,on="Student_id",how="inner")
```

```python
#Left Join
pd.merge(Student,Marks,on="Student_id",how="left")
```

```python
#Right Join
pd.merge(Student,Marks,on="Student_id",how="right")
```

```python
#Outer Join
pd.merge(Student,Marks,on="Student_id",how="outer")
```

```python
# Cross Join
left=pd.DataFrame({"Student":["Amit","Priya"]})
right=pd.DataFrame({"Subject":["Python","Java"]})
pd.merge(left,right,how="cross")
```

```python
# Suffix Join
df2=pd.DataFrame({"Student_id":[101,102],
                  "Marks":[88,89]})
df3=pd.DataFrame({"Student_id":[101,102],
                  "Marks":[76,78]})
pd.merge(df2,df3,on="Student_id",suffixes=("_internal","_external"))
```

```python
r1=pd.merge(Student,Marks,on="Student_id",how="outer",indicator="Source")
print(r1)
```

```python
#one_to_one
df6=pd.DataFrame({"Student_id":[101,103,102],
                  "Name":["Tara","Aayush","Raj"]})
df7=pd.DataFrame({"Student_id":[101],
                  "Marks":[56]})
pd.merge(df6,df7,on="Student_id", validate="one_to_one")
```

```python
#one_to_many
df6=pd.DataFrame({"Student_id":[101,103,102],
                  "Name":["Tara","Aayush","Raj"]})
df7=pd.DataFrame({"Student_id":[101,101],
                  "Marks":[56,44]})
pd.merge(df6,df7,on="Student_id", validate="one_to_many")
```

```python
df=pd.DataFrame({
    "Student":["Amit","Priya","Aayush","Raj","Reena"],
    "Subject":["Java","Python","Java","Python","Java"],
    "Marks":[56,78,67,44,88]
})
print(df)
```

```python
res=df.pivot(index="Student",columns="Subject",values="Marks")
print(res)
```

```python
df=pd.DataFrame({
    "Student":["Amit","Priya","Aayush","Raj","Reena"],
    "Subject":["Java","Java","Java","Java","Java"],
    "Marks":[56,78,67,44,88]
})
print(df)


res1=df.pivot_table(values="Marks",
                    index="Student",
                    columns="Subject",
                    aggfunc="max")
print(res1)
```
