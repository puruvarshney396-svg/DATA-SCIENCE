```python
import pandas as pd
```

```python
#using plot function
df=pd.DataFrame({
    "Department":['CSE','ECE','IT','ME'],
    "Students":[50,40,60,90]
})
df.plot(x='Department',y='Students',kind='line')
```

```python
#using Bar graph & Vertical Wise
df=pd.DataFrame({
    "Department":['CSE','ECE','IT','ME'],
    "Students":[50,40,60,90]
})
df.plot(x='Department',y='Students',kind='bar')
```

```python
#Using Barh function & Horizontal Wise
df=pd.DataFrame({
    "Department":['CSE','ECE','IT','ME'],
    "Students":[50,40,60,90]
})
df.plot(x='Department',y='Students',kind='barh')
```

```python
#using pie chart
df=pd.DataFrame({
    "Department":['CSE','ECE','IT','ME'],
    "Students":[50,40,60,90]
})
df.plot(x='Department',y='Students',kind='pie')
```

# **Program of Sharma Kohli Dataset**

```python
import matplotlib.pyplot as plt
```

```python
df1=pd.read_csv("/content/archive.zip")
df1
```

```python
from matplotlib import legend
plt.plot(df1['index'],df1['V Kohli'],label="Virat Kohli",color='green',marker='o',linestyle='dashed')
plt.plot(df1['index'],df1['RG Sharma'],label="Rohit Sharma",color='red',marker='o',linestyle='-',linewidth=3,markersize=10,)
plt.legend(loc='lower right',fontsize=12)
plt.title("Rohit Sharma vs Virat Kohli")
plt.xlabel('Year',fontsize=14)
plt.ylabel("Run Scored",fontsize=11)
plt.ylim(100,1000)
plt.grid(True,linestyle=":",alpha=0.5)
plt.figure(figsize=(10,5))
plt.savefig("IPL.png")
plt.show()

```
